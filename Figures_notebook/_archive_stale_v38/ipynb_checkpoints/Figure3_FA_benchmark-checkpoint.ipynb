# Figure 3 (FA) Benchmark | Choosing the FactorAnalysis latent dimension

Sweeps the **FactorAnalysis latent dimension** for the bone-marrow `Figure3_FA`
pipeline (DC / Ery / Mono lineages) and picks the most stable, biologically
consistent setting.

**Design**

| | |
|---|---|
| Latent dims | `10, 20, 30` |
| Seeds / setting | `3` (varies FA randomized-SVD, Palantir waypoints, drift-net init) |
| Branches | DC, Ery, Mono (Palantir terminal cells, identical to `Figure3_FA`) |

**Scores (per setting, aggregated over seeds)**

1. **Branch stability** — mean pairwise Jaccard of each branch's top instability
   genes *across seeds*. Higher = more reproducible.
2. **Marker plausibility** — fraction of canonical lineage markers (human panels)
   recovered in each branch's top instability genes. Higher = more biological.
3. **Gene overlap across branches** — mean pairwise Jaccard *between* the three
   branches' gene sets. Lower = more lineage-specific (better separation).

A min-max-normalised **composite** = `stability + marker + (1 − overlap)` selects
the winner.

> **Note:** this is a *screening* run at reduced epochs (`N_EPOCHS=1500`).
> Re-run the chosen dimension at full quality in `Figure3_FA.ipynb`
> (set `N_LATENT = <winner>`, `N_EPOCHS = 5000`).


```python
import sys, os, time, json, warnings
sys.path.insert(0, os.path.abspath('..'))
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import scipy.sparse as sp
import scanpy as sc
import palantir
import scjdo as sjd
from sklearn.decomposition import FactorAnalysis

mpl.rcParams.update({
    'pdf.fonttype': 42, 'ps.fonttype': 42, 'font.family': 'DejaVu Sans',
    'axes.titlesize': 10, 'axes.labelsize': 9,
    'xtick.labelsize': 8, 'ytick.labelsize': 8, 'legend.fontsize': 8,
})

# ── Benchmark grid ──────────────────────────────────────────────────────────
LATENT_DIMS = [10, 20, 30]
SEEDS       = [0, 1, 2]            # 3 seeds per setting
N_TOP_GENES = 25                  # top instability genes scored per branch

# ── Pipeline config (screening epochs; re-run winner at 5000 in Figure3_FA) ──
DATA_PATH    = '../examples/marrow_sample_scseq_counts.h5ad'
N_HVG        = 1500
N_WAYPOINTS  = 500
N_ARCHETYPES = 5
N_EPOCHS     = 1500
BIAS_STRENGTH = 1.5

START_CELL      = 'Run5_164698952452459'
TERMINAL_STATES = pd.Series(
    ['DC', 'Mono', 'Ery'],
    index=['Run5_131097901611291', 'Run5_134936662236454', 'Run4_200562869397916'],
)
BRANCHES      = ['DC', 'Ery', 'Mono']
BRANCH_COLORS = {'Ery': '#E63946', 'DC': '#2A9D8F', 'Mono': '#457B9D'}

# ── Canonical lineage marker panels (human symbols) ─────────────────────────
MARKERS = {
    'Ery':  ['GATA1', 'KLF1', 'HBB', 'HBA1', 'HBA2', 'CA1', 'CA2', 'TFRC', 'GYPA',
             'SLC4A1', 'ALAS2', 'EPOR', 'KEL', 'RHAG', 'SLC25A37', 'BLVRB', 'ZFPM1'],
    'DC':   ['IRF8', 'ID2', 'TCF4', 'SPIB', 'IRF7', 'CD74', 'HLA-DRA', 'HLA-DRB1',
             'CLEC4C', 'IL3RA', 'GZMB', 'BST2', 'PLD4', 'IRF4', 'BATF3', 'FLT3'],
    'Mono': ['MPO', 'ELANE', 'PRTN3', 'CTSG', 'AZU1', 'LYZ', 'CEBPE', 'CEBPD',
             'CEBPA', 'CSF1R', 'CD14', 'FCN1', 'S100A8', 'S100A9', 'MNDA', 'CST3', 'VCAN'],
}

OUTDIR = 'results/figure3_fa_benchmark/'
os.makedirs(OUTDIR, exist_ok=True)

print('scjdo', sjd.__version__, '| palantir', palantir.__version__)
print('Grid:', len(LATENT_DIMS), 'latent dims x', len(SEEDS), 'seeds =',
      len(LATENT_DIMS) * len(SEEDS), 'runs  | screening epochs =', N_EPOCHS)
```

```python
# ── Preprocess ONCE (seed-independent: normalize, log, HVG selection) ───────
base = sc.read(DATA_PATH)
base.layers['raw_counts'] = base.X.copy()
sc.pp.normalize_per_cell(base)
palantir.preprocess.log_transform(base)
sc.pp.highly_variable_genes(base, n_top_genes=N_HVG, flavor='cell_ranger')
print('base:', base.shape, '| HVG:', int(base.var.highly_variable.sum()))

# Only HVG genes receive nonzero FA loadings, so only they can surface as
# instability genes. Restrict marker panels to what is actually recoverable.
all_genes = set(base.var_names)
hvg_set   = set(base.var_names[base.var['highly_variable'].values])
MARKERS_AVAIL = {}
print()
for lin, panel in MARKERS.items():
    in_hvg = [g for g in panel if g in hvg_set]
    MARKERS_AVAIL[lin] = in_hvg
    missing = [g for g in panel if g in all_genes and g not in hvg_set]
    print(f'  {lin:4s}: {len(in_hvg)}/{len(panel)} markers in HVG ->', in_hvg)
    if missing:
        print(f'        (present but not HVG, cannot be recovered: {missing})')
```

```python
def _jaccard(a, b):
    a, b = set(a), set(b)
    u = a | b
    return (len(a & b) / len(u)) if u else np.nan

def marker_fraction(genes, lineage):
    panel = MARKERS_AVAIL[lineage]
    if not panel or not genes:
        return np.nan
    return len(set(genes) & set(panel)) / len(panel)

def run_one(L, seed):
    '''Run the full FA -> Palantir -> branch-drift pipeline once.

    Returns {branch: {r2, n, peak_t, max_eig, genes}}.
    '''
    ad = base.copy()
    np.random.seed(seed)

    # FactorAnalysis latent space (randomized SVD so the seed actually varies it)
    hvg = ad.var['highly_variable'].values
    Xh  = ad[:, hvg].X
    Xh  = Xh.toarray() if sp.issparse(Xh) else np.asarray(Xh, dtype=np.float32)
    fa  = FactorAnalysis(n_components=L, svd_method='randomized',
                         random_state=seed, max_iter=1000)
    ad.obsm['X_fa'] = fa.fit_transform(Xh).astype('float32')
    load = np.zeros((ad.n_vars, L), dtype=np.float32)
    load[hvg] = fa.components_.T.astype('float32')
    ad.varm['PCs'] = load                       # scJDO loadings key (FA loadings here)

    # Diffusion maps + Palantir pseudotime/fate on the FA embedding
    palantir.utils.run_diffusion_maps(ad, n_components=5, pca_key='X_fa')
    palantir.utils.determine_multiscale_space(ad)
    np.random.seed(seed)                         # waypoint sampling
    palantir.core.run_palantir(ad, START_CELL, num_waypoints=N_WAYPOINTS,
                               terminal_states=TERMINAL_STATES)
    palantir.presults.select_branch_cells(ad, q=0.01, eps=0.01,
                                          masks_key='branch_masks', save_as_df=True)

    # Cell-fate labels for the velocity bias (same recipe as Figure3_FA)
    fp = ad.obsm['palantir_fate_probabilities']
    ad.obs['cell_fate'] = fp.idxmax(axis=1).astype(str)
    thr = ad.obs['palantir_entropy'].quantile(0.75)
    ad.obs.loc[ad.obs['palantir_entropy'] >= thr, 'cell_fate'] = 'Progenitor'

    # Branch-separated drift fields in FA space
    sjd.tl.fit_drift_branches(
        ad, rep='X_fa', branch_key='branch_masks',
        branch_names=list(ad.obsm['branch_masks'].columns),
        time_key='palantir_pseudotime', groupby='cell_fate',
        progenitor_cluster='Progenitor',
        terminal_clusters={'DC': 'DC', 'Mono': 'Mono', 'Ery': 'Ery'},
        bias_strength=BIAS_STRENGTH, n_archetypes=N_ARCHETYPES,
        n_epochs=N_EPOCHS, vel_scale=2.0, seed=seed, verbose=False,
    )

    out = {}
    for b in BRANCHES:
        key = f'scjdo_{b}'
        if key not in ad.uns:
            out[b] = dict(r2=np.nan, n=0, peak_t=np.nan, max_eig=np.nan, genes=[])
            continue
        res = ad.uns[key]
        eig = np.asarray(res['max_real_eig'])
        out[b] = dict(
            r2=float(res['r2']), n=int(res['n_cells']),
            peak_t=float(res['t_centers'][int(np.argmax(eig))]) if len(eig) else np.nan,
            max_eig=float(eig.max()) if len(eig) else np.nan,
            genes=list(res.get('top_instability_genes', []))[:N_TOP_GENES],
        )
    return out
```

```python
# ── Run the sweep (this is the slow cell) ──────────────────────────────────
runs    = {}      # (L, seed) -> per-branch dict
records = []       # tidy rows for inspection
t_all = time.time()
for L in LATENT_DIMS:
    for s in SEEDS:
        t0 = time.time()
        try:
            res = run_one(L, s)
            tag = ''
        except Exception as e:
            print(f'[L={L} seed={s}] FAILED: {type(e).__name__}: {e}')
            res = {b: dict(r2=np.nan, n=0, peak_t=np.nan, max_eig=np.nan, genes=[])
                   for b in BRANCHES}
            tag = '  (failed)'
        runs[(L, s)] = res
        r2s = ', '.join(f'{b}={res[b]["r2"]:.3f}' for b in BRANCHES)
        print(f'[L={L:2d} seed={s}]  {time.time()-t0:5.1f}s   R2: {r2s}{tag}')
        for b in BRANCHES:
            d = res[b]
            records.append(dict(latent=L, seed=s, branch=b, r2=d['r2'],
                                n_cells=d['n'], peak_t=d['peak_t'], max_eig=d['max_eig'],
                                n_genes=len(d['genes']), genes=';'.join(d['genes'])))

raw = pd.DataFrame(records)
raw.to_csv(OUTDIR + 'benchmark_raw.csv', index=False)
print()
print(f'Total: {time.time()-t_all:.0f}s   ->  {OUTDIR}benchmark_raw.csv')
```

```python
# ── Aggregate the three scores per setting ─────────────────────────────────
rows = []
for L in LATENT_DIMS:
    # (1) Branch stability: cross-seed reproducibility of each branch's gene set
    stab_b = {}
    for b in BRANCHES:
        sets  = [runs[(L, s)][b]['genes'] for s in SEEDS]
        pairs = [_jaccard(sets[i], sets[j])
                 for i in range(len(sets)) for j in range(i + 1, len(sets))]
        stab_b[b] = np.nanmean(pairs) if pairs else np.nan
    stability = np.nanmean([stab_b[b] for b in BRANCHES])

    # (2) Marker plausibility: mean marker fraction over seeds x branches
    marker = np.nanmean([marker_fraction(runs[(L, s)][b]['genes'], b)
                         for s in SEEDS for b in BRANCHES])

    # (3) Cross-branch overlap: mean pairwise Jaccard between branches, per seed
    cross = []
    for s in SEEDS:
        g = {b: runs[(L, s)][b]['genes'] for b in BRANCHES}
        cross.append(np.nanmean([_jaccard(g['DC'], g['Ery']),
                                 _jaccard(g['DC'], g['Mono']),
                                 _jaccard(g['Ery'], g['Mono'])]))
    overlap = np.nanmean(cross)

    r2_vals = [runs[(L, s)][b]['r2'] for s in SEEDS for b in BRANCHES]
    rows.append(dict(latent=L, stability=stability,
                     stab_DC=stab_b['DC'], stab_Ery=stab_b['Ery'], stab_Mono=stab_b['Mono'],
                     marker_plausibility=marker, cross_branch_overlap=overlap,
                     mean_r2=np.nanmean(r2_vals), std_r2=np.nanstd(r2_vals)))

summary = pd.DataFrame(rows).set_index('latent')

def _mm(x):                                   # min-max normalise across settings
    x = np.asarray(x, float)
    rng = np.nanmax(x) - np.nanmin(x)
    return (x - np.nanmin(x)) / rng if rng > 1e-12 else np.full_like(x, 0.5)

summary['composite'] = (_mm(summary['stability']) +
                        _mm(summary['marker_plausibility']) +
                        (1 - _mm(summary['cross_branch_overlap']))) / 3
summary = summary.sort_values('composite', ascending=False)
summary.to_csv(OUTDIR + 'benchmark_summary.csv')

best = int(summary.index[0])
print('Per-setting scores (sorted by composite; higher = better):')
print(summary.round(3).to_string())
print()
print('  most stable (gene-set reproducibility):', int(summary['stability'].idxmax()), '-D')
print('  most marker-plausible                 :', int(summary['marker_plausibility'].idxmax()), '-D')
print('  lowest cross-branch overlap           :', int(summary['cross_branch_overlap'].idxmin()), '-D')
print()
print(f'==> RECOMMENDED N_LATENT = {best}-D  (best combined stability + biology)')
```

```python
# ── Summary figure (4 panels) ──────────────────────────────────────────────
sumL = summary.sort_index()
x, w = np.arange(len(LATENT_DIMS)), 0.25
fig, axes = plt.subplots(2, 2, figsize=(13, 10))
ax1, ax2, ax3, ax4 = axes.ravel()

# (a) branch stability across seeds
for i, b in enumerate(BRANCHES):
    ax1.bar(x + (i - 1) * w, [sumL.loc[L, f'stab_{b}'] for L in LATENT_DIMS],
            width=w, label=b, color=BRANCH_COLORS[b], alpha=0.85)
ax1.set_xticks(x); ax1.set_xticklabels([f'{L}-D' for L in LATENT_DIMS])
ax1.set_ylabel('Mean pairwise Jaccard (top genes)')
ax1.set_title('a  |  Branch stability across seeds (higher = reproducible)',
              fontweight='bold', loc='left', fontsize=10)
ax1.legend(title='branch', fontsize=8)

# (b) marker plausibility (mean over seeds), grouped by branch
mk = {b: [np.nanmean([marker_fraction(runs[(L, s)][b]['genes'], b) for s in SEEDS])
          for L in LATENT_DIMS] for b in BRANCHES}
for i, b in enumerate(BRANCHES):
    ax2.bar(x + (i - 1) * w, mk[b], width=w, label=b, color=BRANCH_COLORS[b], alpha=0.85)
ax2.set_xticks(x); ax2.set_xticklabels([f'{L}-D' for L in LATENT_DIMS])
ax2.set_ylabel('Fraction of lineage markers recovered')
ax2.set_title('b  |  Marker plausibility (higher = better)',
              fontweight='bold', loc='left', fontsize=10)
ax2.legend(title='branch', fontsize=8)

# (c) cross-branch overlap (lower better)
ax3.bar(x, [sumL.loc[L, 'cross_branch_overlap'] for L in LATENT_DIMS],
        color='#8338EC', alpha=0.85, width=0.5)
ax3.set_xticks(x); ax3.set_xticklabels([f'{L}-D' for L in LATENT_DIMS])
ax3.set_ylabel('Mean cross-branch Jaccard')
ax3.set_title('c  |  Gene overlap across branches (lower = lineage-specific)',
              fontweight='bold', loc='left', fontsize=10)

# (d) composite, winner highlighted
comp = [sumL.loc[L, 'composite'] for L in LATENT_DIMS]
ax4.bar(x, comp, width=0.5, alpha=0.9,
        color=['#F4A261' if L == best else '#adb5bd' for L in LATENT_DIMS])
ax4.set_xticks(x); ax4.set_xticklabels([f'{L}-D' for L in LATENT_DIMS])
ax4.set_ylabel('Composite score')
ax4.set_title(f'd  |  Composite ranking — winner: {best}-D',
              fontweight='bold', loc='left', fontsize=10)
for xi, c in zip(x, comp):
    ax4.text(xi, c + 0.01, f'{c:.2f}', ha='center', fontsize=8)

fig.suptitle('Figure 3 (FA) Benchmark  |  FactorAnalysis latent-dimension sweep '
             '(Bone Marrow: DC / Ery / Mono)', fontsize=12, fontweight='bold')
fig.tight_layout(rect=[0, 0, 1, 0.96])
fig.savefig(OUTDIR + 'Figure3_FA_benchmark.pdf', dpi=300, bbox_inches='tight')
fig.savefig(OUTDIR + 'Figure3_FA_benchmark.png', dpi=200, bbox_inches='tight')
plt.show()
print('Saved ->', OUTDIR + 'Figure3_FA_benchmark.pdf / .png')
```

```python
# ── Persist the recommendation ─────────────────────────────────────────────
recommendation = {
    'recommended_n_latent':        int(best),
    'latent_dims_tested':          LATENT_DIMS,
    'seeds_per_setting':           len(SEEDS),
    'n_epochs_screening':          N_EPOCHS,
    'most_stable':                 int(summary['stability'].idxmax()),
    'most_marker_plausible':       int(summary['marker_plausibility'].idxmax()),
    'lowest_cross_branch_overlap': int(summary['cross_branch_overlap'].idxmin()),
    'scores':                      summary.round(4).reset_index().to_dict(orient='records'),
    'next_step': ('Re-run Figure3_FA.ipynb with N_LATENT=%d and N_EPOCHS=5000 '
                  'for the publication-quality figure.') % best,
}
with open(OUTDIR + 'benchmark_recommendation.json', 'w') as f:
    json.dump(recommendation, f, indent=2)
print(json.dumps(recommendation, indent=2))
```
